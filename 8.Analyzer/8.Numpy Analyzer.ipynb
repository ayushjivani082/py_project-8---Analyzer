## Numpy Analyzer

```python
import numpy as np
```

```python
class DataAnalytics:

    def __init__(self):
        self.__array = None

    def __get_array(self):
        return self.__array

    def __set_array(self, array):
        self.__array = np.array(array)

    def __check_array(self):
        if self.__array is None:
            print("Please create an array first!")
            return False
        return True

    def read_array(self, size):
        try:
            values = list(map(float, input("Enter elements separated by space: ").split()))

            if len(values) != size:
                print("Please enter exactly", size, "elements.")
                return None

            return np.array(values)

        except ValueError:
            print("Invalid input!")
            return None

    def show_array(self):
        if self.__check_array():
            print("\nCurrent Array:")
            print(self.__get_array())
            return True
        return False

    @classmethod
    def project_name(cls):
        return "NUMPY ANALYZER"

    @staticmethod
    def line():
        print("=" * 50)
```

```python
def create_array(self):

    print("\n1. 1D Array")
    print("2. 2D Array")
    print("3. 3D Array")

    choice = input("Enter choice: ")

    try:
        if choice == "1":
            size = int(input("Enter size: "))

            if size <= 0:
                print("Size must be positive.")
                return

            array = self.read_array(size)

        elif choice == "2":
            rows = int(input("Enter rows: "))
            cols = int(input("Enter columns: "))

            if rows <= 0 or cols <= 0:
                print("Invalid dimensions.")
                return

            array = self.read_array(rows * cols)

            if array is not None:
                array = array.reshape(rows, cols)

        elif choice == "3":
            blocks = int(input("Enter blocks: "))
            rows = int(input("Enter rows: "))
            cols = int(input("Enter columns: "))

            if min(blocks, rows, cols) <= 0:
                print("Invalid dimensions.")
                return

            array = self.read_array(blocks * rows * cols)

            if array is not None:
                array = array.reshape(blocks, rows, cols)

        else:
            print("Invalid choice.")
            return

        if array is not None:
            self.__set_array(array)

            print("\nArray Created Successfully!")
            print(self.__get_array())
            print("Dimensions:", array.ndim)
            print("Shape:", array.shape)

    except ValueError:
        print("Please enter valid dimensions.")


DataAnalytics.create_array = create_array
```

```python
def indexing_slicing(self):

    if not self.show_array():
        return

    array = self._DataAnalytics__get_array()

    print("\n1. Indexing")
    print("2. Slicing")

    choice = input("Enter choice: ")

    try:
        if choice == "1":

            indexes = []

            for i in range(array.ndim):
                indexes.append(int(input("Enter index for dimension " + str(i + 1) + ": ")))

            print("Selected Element:", array[tuple(indexes)])

        elif choice == "2":

            slices = []

            for i in range(array.ndim):
                start = int(input("Enter start for dimension " + str(i + 1) + ": "))
                end = int(input("Enter end for dimension " + str(i + 1) + ": "))
                slices.append(slice(start, end))

            print("Sliced Array:")
            print(array[tuple(slices)])

        else:
            print("Invalid choice.")

    except (ValueError, IndexError):
        print("Invalid index or slicing range.")


DataAnalytics.indexing_slicing = indexing_slicing
```

```python
def mathematical_operations(self):

    if not self.show_array():
        return

    array = self._DataAnalytics__get_array()

    print("\n1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Dot Product")
    print("6. Matrix Multiplication")

    choice = input("Enter choice: ")

    if choice not in ["1", "2", "3", "4", "5", "6"]:
        print("Invalid choice.")
        return

    try:
        if choice in ["5", "6"]:

            if array.ndim != 2:
                print("Please use a 2D array.")
                return

            rows = array.shape[1]
            cols = int(input("Enter second array columns: "))

            if cols <= 0:
                print("Invalid columns.")
                return

            second = self.read_array(rows * cols)

            if second is None:
                return

            second = second.reshape(rows, cols)

        else:
            second = self.read_array(array.size)

            if second is None:
                return

            second = second.reshape(array.shape)

        if choice == "1":
            result = array + second
            print("Addition:", result)

        elif choice == "2":
            result = array - second
            print("Subtraction:", result)

        elif choice == "3":
            result = array * second
            print("Multiplication:", result)

        elif choice == "4":
            if np.any(second == 0):
                print("Cannot divide by zero.")
                return

            result = array / second
            print("Division:", result)

        elif choice == "5":
            result = np.dot(array, second)
            print("Dot Product:")
            print(result)

        else:
            result = np.matmul(array, second)
            print("Matrix Multiplication:")
            print(result)

    except ValueError as e:
        print("Operation error:", e)


DataAnalytics.mathematical_operations = mathematical_operations
```

```python
def combine_split(self):

    if not self.show_array():
        return

    array = self._DataAnalytics__get_array()

    print("\n1. Combine Arrays")
    print("2. Split Array")

    choice = input("Enter choice: ")

    try:
        if choice == "1":

            if array.ndim == 1:
                second = self.read_array(array.size)

                if second is None:
                    return

                result = np.concatenate((array, second))

            else:
                print("1. Horizontal")
                print("2. Vertical")

                option = input("Enter choice: ")

                axis = array.ndim - 1 if option == "1" else array.ndim - 2 if option == "2" else -1

                if axis == -1:
                    print("Invalid choice.")
                    return

                shape = list(array.shape)
                size = 1

                for i in range(array.ndim):
                    if i != axis:
                        size *= shape[i]

                second = self.read_array(size)

                if second is None:
                    return

                second_shape = list(shape)
                second_shape[axis] = 1
                second = second.reshape(second_shape)

                result = np.concatenate((array, second), axis=axis)

            print("Combined Array:")
            print(result)

        elif choice == "2":

            axis = 0

            if array.ndim > 1:
                axis = int(input("Enter axis: "))

            if axis < 0 or axis >= array.ndim:
                print("Invalid axis.")
                return

            parts = int(input("Enter number of parts: "))

            if parts <= 0:
                print("Invalid number of parts.")
                return

            result = np.split(array, parts, axis=axis)

            print("Split Arrays:")
            for part in result:
                print(part)

        else:
            print("Invalid choice.")

    except ValueError as e:
        print("Combine/Split error:", e)


DataAnalytics.combine_split = combine_split
```

```python
def search_sort_filter(self):

    if not self.show_array():
        return

    array = self._DataAnalytics__get_array()

    print("\n1. Search")
    print("2. Ascending Sort")
    print("3. Descending Sort")
    print("4. Filtering")

    choice = input("Enter choice: ")

    try:
        if choice == "1":

            value = float(input("Enter value to search: "))

            positions = np.argwhere(array == value)

            if positions.size:
                print("Value found at:")
                print(positions)
            else:
                print("Value not found.")

        elif choice == "2":
            print("Ascending Order:")
            print(np.sort(array, axis=None))

        elif choice == "3":
            print("Descending Order:")
            print(np.sort(array, axis=None)[::-1])

        elif choice == "4":

            print("\n1. Greater Than")
            print("2. Less Than")
            print("3. Equal To")

            option = input("Enter condition: ")
            value = float(input("Enter value: "))

            if option == "1":
                result = array[array > value]

            elif option == "2":
                result = array[array < value]

            elif option == "3":
                result = array[array == value]

            else:
                print("Invalid choice.")
                return

            print("Filtered Values:")
            print(result)

        else:
            print("Invalid choice.")

    except ValueError:
        print("Invalid input.")


DataAnalytics.search_sort_filter = search_sort_filter
```

```python
def aggregating_functions(self):

    if not self.show_array():
        return

    array = self._DataAnalytics__get_array()

    print("\nAGGREGATING FUNCTIONS")

    print("Sum:", np.sum(array))

    print("Mean:", np.mean(array))

    print("Median:", np.median(array))

    print("Standard Deviation:", np.std(array))

    print("Variance:", np.var(array))


DataAnalytics.aggregating_functions = aggregating_functions
```

```python
def statistical_functions(self):

    if not self.show_array():
        return

    array = self._DataAnalytics__get_array()

    print("\nSTATISTICAL ANALYSIS")

    print("Minimum:", np.min(array))

    print("Maximum:", np.max(array))

    percentile = float(input("Enter percentile (0-100): "))

    if percentile < 0 or percentile > 100:
        print("Invalid percentile.")
        return

    print("Percentile:", np.percentile(array, percentile))

    second = self.read_array(array.size)

    if second is None:
        return

    second = second.reshape(array.shape)

    first = array.flatten()
    second = second.flatten()

    if first.size < 2:
        print("At least two elements are required.")

    elif np.std(first) == 0 or np.std(second) == 0:
        print("Correlation cannot be calculated.")

    else:
        correlation = np.corrcoef(first, second)[0, 1]

        print("Correlation:", correlation)


DataAnalytics.statistical_functions = statistical_functions
```

```python
def main_menu(self):

    print("\n")
    self.line()
    print("       WELCOME TO NUMPY ANALYZER")
    self.line()

    while True:

        print("\n1. Create NumPy Array")
        print("2. Indexing and Slicing")
        print("3. Mathematical Operations")
        print("4. Combine and Split Arrays")
        print("5. Search, Sort and Filter")
        print("6. Aggregating Functions")
        print("7. Statistical Functions")
        print("8. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            self.create_array()

        elif choice == "2":
            self.indexing_slicing()

        elif choice == "3":
            self.mathematical_operations()

        elif choice == "4":
            self.combine_split()

        elif choice == "5":
            self.search_sort_filter()

        elif choice == "6":
            self.aggregating_functions()

        elif choice == "7":
            self.statistical_functions()

        elif choice == "8":
            print("\nThank you for using NumPy Analyzer!")
            break

        else:
            print("Invalid choice! Try again.")


DataAnalytics.main_menu = main_menu
```

```python
analyzer = DataAnalytics()

analyzer.main_menu()
```
